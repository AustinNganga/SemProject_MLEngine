# RandomForest_Final — MindTrack ML Engine

## Predicting Depression (Beck Score) from Phone-Use, Background and Anxiety Questionnaires

**Final ML approach: Random forest** (300 decision trees, `min_samples_leaf=5`, balanced class weights) on 20 questionnaire inputs, with a screening threshold chosen for ≈75% sensitivity. The trained model is exported to JSON and runs on the phone in the MindTrack Android app (repository `MindTrack_mobile_app`).

**End-to-end pipeline:** raw data → cleaning → EDA → train/test split → feature-set comparison → PCA → model comparison & tuning → compact app question set → decision threshold → held-out test → Beck-score regression → mobile export (JSON + Kotlin) → parity check.

Everything runs inside this notebook; no external scripts are called. Run top-to-bottom (Kernel → Restart & Run All). Runtime is roughly 3–6 minutes.

| | |
|---|---|
| **Dataset** | *Digital Sedentary Behavior, Mobile Device Dependence* (1,220 university students, 5 linked questionnaires) |
| **Outcome** | Beck Depression Inventory total (0–63). App output = risk that Beck total ≥ 11 (codebook: *mild mood disturbance* or worse) |
| **Main result** | Random forest on a short app form. CV ROC-AUC ≈ 0.75, PR-AUC ≈ 0.42 (no-skill 0.18). Test ROC-AUC ≈ 0.76 |

> **Ethics note.** The model estimates a *questionnaire score group*. It is a screening aid, **not a diagnosis**. The app must say so and should point users to support services.

## Key terms in plain words

| Term | Plain-language meaning |
|---|---|
| **Feature / input** | One piece of information the model uses, e.g. one questionnaire answer or the student's age. |
| **Target / outcome** | What we want to predict: whether the Beck depression score is 11 or more (`Beck_11plus`). |
| **Base model (baseline)** | The simplest possible "model" to beat. Ours always predicts the most common answer ("not depressed"). Any useful model must do better than it. |
| **Train / test split** | We hide 20% of students (the *test set*) and only look at them once at the very end, to get an honest score on people the model has never seen. |
| **Cross-validation (CV)** | Repeatedly splitting the *training* data into 5 parts, training on 4 and checking on the 5th, so that model choices are not based on luck. "Repeated" means we do this 3 times with different splits. |
| **Overfitting** | When a model memorises the training data instead of learning general patterns, so it does worse on new people. CV and the test set catch this. |
| **Decision tree** | A flowchart of yes/no questions (e.g. "Is the anxiety answer above 2?") that ends in a prediction. |
| **Random forest** | 300 decision trees, each trained on a slightly different sample of students. Their answers are averaged, which is more reliable than one tree. |
| **Gradient boosting** | Trees built one after another, each fixing the previous ones' mistakes. We tried it; the random forest did better here. |
| **Logistic regression** | A simpler model that adds up weighted answers. It is easy to interpret but cannot capture complex patterns. |
| **Probability / risk score** | The model's estimate (0 to 1) that a student is in the Beck ≥ 11 group. The app shows it on a 0–100 scale. |
| **Threshold (cut-off)** | The score above which we call a result "Elevated". We chose it so the model catches about 75% of struggling students. |
| **Recall (sensitivity)** | Of the students who really scored 11+, the share the model flagged. High recall = few missed students. |
| **Specificity** | Of the students who scored below 11, the share the model correctly called "Low". |
| **Precision** | Of the students the model flagged, the share who really scored 11+. |
| **ROC-AUC** | How well the model ranks students: the chance that a random struggling student gets a higher score than a random non-struggling one. 0.5 = guessing, 1.0 = perfect. |
| **PR-AUC (average precision)** | Like ROC-AUC but focused on the smaller "struggling" group, which makes it more honest when that group is rare (18% here). Guessing scores about 0.18. |
| **Class imbalance** | One outcome is much rarer than the other (18% vs 82%). We use `class_weight='balanced'` so the model doesn't just ignore the rare group. |
| **PCA (principal component analysis)** | A way to squeeze many related columns into a few summary columns. We tested it; it made predictions worse, so it is not used. |
| **Cronbach's alpha** | How consistently the items of a questionnaire measure the same thing (0.7+ is usually acceptable). |
| **Odds ratio (OR) and p-value** | From the hypothesis test: OR > 1 means higher phone dependence goes with higher odds of the Beck ≥ 11 group; p < 0.05 means this is unlikely to be chance. |
| **Permutation importance** | How much the test score drops when one input is shuffled. A big drop means the model relies on that input. |
| **Explainable AI (XAI)** | Showing *why* a model gave a particular result. The app splits each prediction into how much each answer pushed the score up or down (tree-path contributions). |
| **Parity check** | Proof that the phone's copy of the model gives the same numbers as the Python model. |

## 1. Problem definition

* **Question:** Can information a student can enter in a phone app (background, phone habits, phone-dependence and anxiety questionnaires) estimate their depression score?
* **Target used for the app:** `Beck_11plus` = 1 if Beck total ≥ 11. The data dictionary labels 0–10 *normal ups and downs*, 11–16 *mild mood disturbance*, 17–20 *borderline*, 21+ *moderate/severe*. 11 is therefore the first band above "normal". This is a natural screening cut-off.
* **Secondary target:** the continuous Beck total (regression), Section 11. Section 11 shows why the app reports a risk group rather than an exact score.
* **Never used as inputs:** the 21 Beck item answers, the Beck sum and level (they *are* the outcome), and the student `Code` (identifier; it is also correlated with the outcome because of collection order, Section 4).

## 2. Load and align the raw questionnaires

```python
files = {
    'socio':  ('2. Dataset_Sociodemo.xlsx', 'SocioDemo_data'),
    'tmd':    ('3. Dataset_MobileDependence.xlsx', 'TMD_data'),
    'nordic': ('4. Dataset_MusculoskeletalSymptoms.xlsx', 'Nordic_Questionnaire_data'),
    'beck':   ('5. Dataset_Depression.xlsx', 'Depression_data'),
    'stai':   ('6. Dataset_Anxiety.xlsx', 'Anxiety_data'),
}
raw = {k: pd.read_excel(RAW / f, sheet_name=s) for k, (f, s) in files.items()}

# Column headers contain stray spaces and the section title glued to the first item
# (e.g. 'TMD                    Q1'). Keep only the last token for those, strip the rest.
def clean_header(c):
    c = str(c).strip()
    return c.split()[-1] if c.split()[0] in {'TMD', 'Nordic', 'State_Anxiety', 'Trait_Anxiety'} else c
for k in raw:
    raw[k].columns = [clean_header(c) for c in raw[k].columns]
raw['beck'] = raw['beck'].rename(columns={'DepressionA': 'A'})

inv = pd.DataFrame([{'file': files[k][0], 'rows': len(d), 'columns': d.shape[1],
                     'unique_codes': d['Code'].nunique(), 'missing_cells': int(d.isna().sum().sum()),
                     'duplicate_rows': int(d.drop(columns='Code').duplicated().sum())} for k, d in raw.items()])
display(inv)
codes = raw['socio']['Code']
assert all(raw[k]['Code'].equals(codes) for k in raw), 'Codes not aligned across files'
print(f'All {len(codes)} student codes are unique and aligned row-by-row in all five files.')
```

## 3. Data cleaning

The audit checks and fixes the following:
1. **Consistent names.** Headers have typos and spaces (`Network _Access`, `Using_Ttime-management_Apps`, `Usual_Phone_Locationn`). They are renamed to clean snake_case so the app uses the same field names.
2. **Codebook validity.** Every coded answer is checked against the data dictionary's allowed range.
3. **Recomputed scores.** The TMD total is recomputed from its 22 items, and the Beck total is checked against its 21 items.
4. **STAI scoring problem.** The supplied anxiety totals are raw sums. Standard STAI scoring reverse-scores the positively worded items (*"I feel calm"*), so we recompute correctly scored totals. The model uses the 40 item answers directly.
5. **Binary recoding.** App-use flags are recoded from 1/2 to 0/1.
6. **Sparse categories.** These are documented, not deleted: daily use "1–3 h" has only 1 student, and some marital-status codes have only a few.
7. **No rows removed.** There are no missing values or duplicate IDs. High scores are valid answers, not outliers, so all 1,220 students are kept.

```python
s = raw['socio'].rename(columns={
    'MS': 'Marital_Status', 'Network _Access': 'Network_Access', 'Usual_Phone_Locationn': 'Usual_Phone_Location',
    'Using_Text Messaging_Apps': 'App_Messaging', 'Using_Games_Apps': 'App_Games',
    'Using_social media_Apps': 'App_Social_Media', 'Using_Music_Apps': 'App_Music',
    'Using_Videos_Apps': 'App_Videos', 'Using_Ttime-management_Apps': 'App_Time_Management',
    'Using_Camara_Apps': 'App_Camera', 'Using_Mail_Apps': 'App_Mail'})
s.columns = [c.strip().replace(' ', '_').replace("'", '') for c in s.columns]

# Allowed ranges taken from '1. Data_Dictionary.xlsx'
valid = {'Department': (1, 4), 'Semester': (1, 10), 'Age': (16, 60), 'Sex': (1, 2), 'Marital_Status': (1, 6),
         'SES': (1, 6), 'Living_Arrangement': (1, 6), 'Family_Expression': (1, 4), 'Family_Activities': (1, 4),
         'Mobile_Use_Rules': (1, 4), 'Phone_Discrimination': (1, 4), 'Phone_Related_Arguments': (1, 4),
         'Residence_Area': (1, 2), 'Subscription_Plan': (1, 2), 'Network_Access': (1, 3), 'Areas_Use': (1, 4),
         'Time_First_Cellphone': (1, 5), 'Type_MD': (1, 4), 'Average_Time_Use': (1, 4)}
valid.update({c: (1, 2) for c in s.columns if c.startswith('App_')})
tmd_items = [f'Q{i}' for i in range(1, 23)]
beck_items = list('ABCDEFGHIJKLMNOPQRSTU')
stai_items = [f'item_{i}' for i in range(1, 41)]

rows = []
for col, (lo, hi) in valid.items():
    v = pd.to_numeric(s[col], errors='coerce'); rows.append((f'socio.{col}', lo, hi, int(((v < lo) | (v > hi) | v.isna()).sum())))
for src, items, lo, hi in [('tmd', tmd_items, 0, 4), ('beck', beck_items, 0, 3), ('stai', stai_items, 1, 4)]:
    v = raw[src][items].apply(pd.to_numeric, errors='coerce')
    rows.append((f'{src} items', lo, hi, int(((v < lo) | (v > hi) | v.isna()).sum().sum())))
validity = pd.DataFrame(rows, columns=['field', 'min_allowed', 'max_allowed', 'invalid_values'])
display(validity[validity.invalid_values > 0] if (validity.invalid_values > 0).any() else
        Markdown('**All coded answers are within the codebook ranges.**'))
```

```python
tmd = raw['tmd'][tmd_items].astype(int)
beck = raw['beck'][beck_items].astype(int)
stai = raw['stai'][stai_items].astype(int)

tmd_total = tmd.sum(axis=1)
print('TMD: supplied totals that disagree with the 22 items:', int((raw['tmd']['sum'] != tmd_total).sum()), '-> recomputed total used')
beck_total = beck.sum(axis=1)
assert (beck_total == raw['beck']['Sum']).all()
print('Beck: supplied total equals the sum of the 21 items for every student.')

# STAI: standard positively-worded (reverse-scored) items
REV_STATE = [1, 2, 5, 8, 10, 11, 15, 16, 19, 20]
REV_TRAIT = [21, 23, 26, 27, 30, 33, 34, 36, 39]
def stai_total(items, rev):
    return sum((5 - stai[f'item_{i}']) if i in rev else stai[f'item_{i}'] for i in items)
state_correct = stai_total(range(1, 21), REV_STATE)
trait_correct = stai_total(range(21, 41), REV_TRAIT)
print('STAI: supplied State total == plain unreversed sum for', f"{(raw['stai']['SUM_State_Anxiety'] == stai.iloc[:, :20].sum(axis=1)).mean():.0%}",
      'of students -> positive items were NOT reverse-scored. Correct totals recomputed.')

sparse = {c: s[c].value_counts().loc[lambda x: x < 10].to_dict() for c in valid if c.startswith(('Average', 'Marital', 'Type_MD', 'Living'))}
print('Sparse categories (<10 students) kept but noted:', {k: v for k, v in sparse.items() if v})
```

```python
socio_cols = ['Age', 'Semester', 'Department', 'Sex', 'Marital_Status', 'SES', 'Living_Arrangement',
              'Family_Expression', 'Family_Activities', 'Mobile_Use_Rules', 'Phone_Discrimination',
              'Phone_Related_Arguments', 'Residence_Area', 'Subscription_Plan', 'Network_Access', 'Areas_Use',
              'Time_First_Cellphone', 'Type_MD', 'Average_Time_Use'] + [c for c in s.columns if c.startswith('App_')]
posture_cols = [c for c in s.columns if c.startswith(('Body_Position', 'Usage_time'))] + \
               ['Hand_Dominance', 'Usual_Phone_Location', 'Primary_input_finger', 'Frequent_touch_gesture']

df = s[socio_cols + posture_cols].copy()
for c in [c for c in socio_cols if c.startswith('App_')]:
    df[c] = (df[c] == 2).astype(int)                       # 1/2 -> 0/1
df[[f'TMD_{q}' for q in tmd_items]] = tmd.values
df['TMD_Total'] = tmd_total
nordic_cols = [f'NOR_{c}' for c in raw['nordic'].columns[8:]]
df[nordic_cols] = raw['nordic'].iloc[:, 8:].values
df[[f'STAI_{q}' for q in stai_items]] = stai.values
df['STAI_State_Correct'] = state_correct
df['STAI_Trait_Correct'] = trait_correct
df['Beck_Total'] = beck_total
df['Beck_11plus'] = (beck_total >= 11).astype(int)

assert df.isna().sum().sum() == 0
df.to_csv(ROOT / 'data/processed/mindtrack_clean.csv', index=False)
print('Clean analysis table:', df.shape, '-> data/processed/mindtrack_clean.csv')
print(f"Beck >= 11: {df.Beck_11plus.sum()} of {len(df)} students ({df.Beck_11plus.mean():.1%})")

NOMINAL = ['Department', 'Sex', 'Marital_Status', 'Living_Arrangement', 'Residence_Area', 'Subscription_Plan',
           'Network_Access', 'Areas_Use', 'Type_MD', 'Hand_Dominance', 'Usual_Phone_Location',
           'Primary_input_finger', 'Frequent_touch_gesture']
```

### 4.4 Reliability of the questionnaires (Cronbach's alpha)

Cronbach's alpha measures how consistently the items of a scale measure the same thing. About 0.70 or higher is usually considered acceptable. STAI alpha is computed both with correct reverse-scoring and with the supplied (unreversed) coding.

```python
def cronbach_alpha(items):
    X = np.asarray(items, float); k = X.shape[1]
    return k / (k - 1) * (1 - X.var(axis=0, ddof=1).sum() / X.sum(axis=1).var(ddof=1))
stai_scored = pd.DataFrame({i: (5 - stai[f'item_{i}']) if i in REV_STATE + REV_TRAIT else stai[f'item_{i}'] for i in range(1, 41)})
rel = pd.DataFrame([
    ('TMD phone dependence', 22, cronbach_alpha(tmd)),
    ('Beck Depression Inventory', 21, cronbach_alpha(beck)),
    ('STAI state, correctly reverse-scored', 20, cronbach_alpha(stai_scored[list(range(1, 21))])),
    ('STAI trait, correctly reverse-scored', 20, cronbach_alpha(stai_scored[list(range(21, 41))])),
    ('STAI state, supplied coding', 20, cronbach_alpha(stai.iloc[:, :20])),
    ('STAI trait, supplied coding', 20, cronbach_alpha(stai.iloc[:, 20:])),
], columns=['scale', 'items', 'cronbach_alpha'])
rel.to_csv(ROOT / 'results/reliability_cronbach_alpha.csv', index=False)
display(rel.round(3))
```

**Reading:** TMD and Beck are highly reliable (α > 0.9). STAI state is good once scored correctly. STAI trait is acceptable (≈ 0.70). Correct reverse-scoring raises the state scale's alpha, which supports the scoring fix in Section 3.

### 4.5 Hypothesis test: adjusted association between phone dependence and the Beck ≥ 11 group

* **H₀:** After adjusting for the prespecified covariates, the recalculated TMD total is not associated with the odds of Beck ≥ 11.
* **H₁:** After adjustment, a higher TMD total is associated with higher odds of Beck ≥ 11.

This is a multivariable logistic regression on all 1,220 students: an inferential analysis, separate from the predictive models. The TMD total is scaled per 10 points. Covariates: age, sex, semester, socioeconomic stratum, department, average daily phone use and time since first phone. Two-sided test at α = 0.05.

```python
import statsmodels.formula.api as smf
assoc = df.assign(TMD_per10=df.TMD_Total / 10, Sex=s['Sex'], SES=s['SES'], Department=s['Department'])
formula = 'Beck_11plus ~ TMD_per10 + Age + C(Sex) + Semester + SES + C(Department) + Average_Time_Use + Time_First_Cellphone'
logit_adj = smf.logit(formula, assoc).fit(disp=0)
logit_unadj = smf.logit('Beck_11plus ~ TMD_per10', assoc).fit(disp=0)
def or_row(m, name, term='TMD_per10'):
    lo, hi = np.exp(m.conf_int().loc[term])
    return {'model': name, 'odds_ratio_per_10_TMD_points': np.exp(m.params[term]), 'ci95_low': lo, 'ci95_high': hi, 'p_value': m.pvalues[term], 'n': int(m.nobs)}
hyp = pd.DataFrame([or_row(logit_unadj, 'Unadjusted'), or_row(logit_adj, 'Adjusted (prespecified covariates)')])
hyp.to_csv(ROOT / 'results/hypothesis_test_logistic.csv', index=False)
display(hyp.round(4))
full = pd.DataFrame({'odds_ratio': np.exp(logit_adj.params), 'ci95_low': np.exp(logit_adj.conf_int()[0]), 'ci95_high': np.exp(logit_adj.conf_int()[1]), 'p_value': logit_adj.pvalues})
display(full.round(3))
decision = 'reject H0' if logit_adj.pvalues['TMD_per10'] < 0.05 else 'fail to reject H0'
print(f'Decision at the 5% level: {decision}')
```

**Reading:** Each extra 10 points on the TMD (range 0–88) is associated with about 16% higher odds of being in the Beck ≥ 11 group, after adjustment. The confidence interval excludes 1, so **H₀ is rejected**. The effect is modest, and because the data are cross-sectional this is an association, not evidence that phone dependence causes low mood. Younger age and the Engineering department were also associated with higher odds (see the full table); Section 6.1 explains why department is not used in the app.

## 5. Train/test split (done *before* any modelling decision)

From here on, every decision (feature groups, PCA, model choice, hyper-parameters, compact question set, threshold) uses **only the 80% training set**, through repeated stratified 5-fold cross-validation. The 20% test set is touched **once**, in Section 10. Keeping it untouched until then avoids the optimistic bias of tuning on test data.

```python
TARGET, SCORE = 'Beck_11plus', 'Beck_Total'
train_df, test_df = train_test_split(df, test_size=0.20, stratify=df[TARGET], random_state=SEED)
train_df.to_csv(ROOT / 'data/splits/train_80.csv', index=False); test_df.to_csv(ROOT / 'data/splits/test_20.csv', index=False)
ytr, yte = train_df[TARGET].values, test_df[TARGET].values
print(f'Train: {len(train_df)} ({ytr.sum()} positive, {ytr.mean():.1%}) | Test: {len(test_df)} ({yte.sum()} positive, {yte.mean():.1%})')

CV = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=SEED)

def preprocess(cols, scale=True):
    nom = [c for c in cols if c in NOMINAL]; num = [c for c in cols if c not in NOMINAL]
    return ColumnTransformer([('onehot', OneHotEncoder(handle_unknown='ignore', min_frequency=10), nom),
                              ('num', StandardScaler() if scale else 'passthrough', num)])

def cv_eval(model, cols, name, **extra):
    r = cross_validate(model, train_df[cols], ytr, cv=CV, scoring=['roc_auc', 'average_precision', 'balanced_accuracy'], n_jobs=-1)
    return {'name': name, **extra, 'n_inputs': len(cols),
            'cv_roc_auc': r['test_roc_auc'].mean(), 'cv_roc_auc_sd': r['test_roc_auc'].std(),
            'cv_pr_auc': r['test_average_precision'].mean(), 'cv_pr_auc_sd': r['test_average_precision'].std()}

def rf(n=300, leaf=5):
    return RandomForestClassifier(n_estimators=n, min_samples_leaf=leaf, max_features='sqrt',
                                  class_weight='balanced_subsample', random_state=SEED, n_jobs=1)
```

## 6. Which questionnaires carry signal? (feature-group comparison, training CV only)

We add one data source at a time, using the same random forest each time, to see which sources actually help.

```python
BACKGROUND = socio_cols
TMD_ITEMS = [f'TMD_{q}' for q in tmd_items]
STAI_ITEMS = [f'STAI_{q}' for q in stai_items]
groups = {
    '1. Background + phone habits':                 BACKGROUND,
    '2. + TMD total (previous project set)':        BACKGROUND + ['TMD_Total'],
    '3. + 22 TMD items':                            BACKGROUND + ['TMD_Total'] + TMD_ITEMS,
    '4. + posture/ergonomics':                      BACKGROUND + ['TMD_Total'] + posture_cols,
    '5. + Nordic musculoskeletal':                  BACKGROUND + ['TMD_Total'] + nordic_cols,
    '6. + STAI totals (correctly scored)':          BACKGROUND + ['TMD_Total', 'STAI_State_Correct', 'STAI_Trait_Correct'],
    '7. + 40 STAI items':                           BACKGROUND + ['TMD_Total'] + STAI_ITEMS,
    '8. Everything':                                BACKGROUND + ['TMD_Total'] + TMD_ITEMS + posture_cols + nordic_cols + STAI_ITEMS,
}
t0 = time.time()
fg = pd.DataFrame([cv_eval(make_pipeline(preprocess(c, scale=False), rf()), c, k) for k, c in groups.items()])
display(fg.round(3)); print(f'{time.time()-t0:.0f}s')
fig, ax = plt.subplots(figsize=(9, 4))
ax.barh(fg.name, fg.cv_pr_auc, xerr=fg.cv_pr_auc_sd, color='#4C72B0'); ax.axvline(ytr.mean(), color='crimson', ls='--', label='no-skill PR-AUC')
ax.invert_yaxis(); ax.set(xlabel='CV PR-AUC (higher is better)', title='Random forest by feature group'); ax.legend()
savefig('06_feature_group_comparison')
FEATURES_WITH_DEPT = groups['7. + 40 STAI items']
```

**Decision:** **Anxiety (STAI) answers are the one source that clearly improves prediction.** PR-AUC rises from about 0.27 to about 0.42. Even the correctly scored STAI totals, which had almost no *linear* correlation with Beck, help a lot once the forest can combine them non-linearly with the other fields. The 40 item answers do slightly better again on ROC-AUC. TMD items, posture and Nordic fields add little or nothing beyond the TMD total, and in an app every extra field is another question to answer. We keep **background + TMD total + STAI items** (`FEATURES`). This also fits the app, because every field is something a student can answer on their phone.

### 6.1 Generalisability decision: remove `Department` and `SES`

`Department` only has four options (Law, Economics, Engineering, Health) from the one Colombian university that collected the data. Its relationship with depression (Engineering about twice the rate of Law and Economics) is institution-specific and partly reflects how data was collected. A student at another university would have to pick the "closest" faculty, and their risk would move for reasons unrelated to how they feel. `SES` has the same problem: it uses Colombia's six-level *estrato* housing-stratum system, which students elsewhere cannot answer meaningfully. We test what removing them costs.

```python
FEATURES = [f for f in FEATURES_WITH_DEPT if f not in ('Department', 'SES')]
dept_cmp = pd.DataFrame([cv_eval(rf(), FEATURES_WITH_DEPT, 'With Department and SES'), cv_eval(rf(), FEATURES, 'Without Department and SES')])
display(dept_cmp.round(3))
display(train_df.groupby('Department')[TARGET].agg(['mean', 'size']).rename(index={1: 'Law', 2: 'Economics', 3: 'Engineering', 4: 'Health'}).round(2))
print(f'Final candidate inputs: {len(FEATURES)}')
```

**Decision:** The difference is well within the cross-validation standard deviation, so **`Department` and `SES` are removed**. The model loses almost nothing and becomes fairer and more transferable. `Age` and `Semester` stay because they mean the same thing at any university.

## 7. PCA

PCA is unsupervised. It finds directions of maximum *predictor* variance, which need not be the directions that predict depression. We use it in two ways:
1. **Exploration** (fitted on the training set only): how much structure is there, and do the two groups separate?
2. **As a model step inside the CV pipeline**, to test whether compressing the inputs helps prediction.

```python
pca_pre = preprocess(FEATURES, scale=True)
Ztr = pca_pre.fit_transform(train_df[FEATURES])
pca_full = PCA(random_state=SEED).fit(Ztr)
cum = np.cumsum(pca_full.explained_variance_ratio_)
k80 = int(np.argmax(cum >= 0.80) + 1)
scores = pca_full.transform(Ztr)[:, :2]
fig, ax = plt.subplots(1, 2, figsize=(13, 4.3))
ax[0].bar(range(1, 31), pca_full.explained_variance_ratio_[:30], color='#8DA0CB', label='per component')
ax[0].plot(range(1, 31), cum[:30], 'o-', color='#C44E52', label='cumulative'); ax[0].axhline(.8, ls=':', color='grey')
ax[0].set(title=f'Scree plot — {k80} of {Ztr.shape[1]} components needed for 80%', xlabel='component'); ax[0].legend()
for lab_, c in [(0, '#4C72B0'), (1, '#DD8452')]:
    m = ytr == lab_; ax[1].scatter(scores[m, 0], scores[m, 1], s=10, alpha=.5, c=c, label='Beck ≥ 11' if lab_ else 'Beck < 11')
ax[1].set(title='Training students on PC1–PC2', xlabel=f'PC1 ({pca_full.explained_variance_ratio_[0]:.1%})', ylabel=f'PC2 ({pca_full.explained_variance_ratio_[1]:.1%})'); ax[1].legend()
savefig('07_pca')
names = pca_pre.get_feature_names_out()
load = pd.DataFrame(pca_full.components_[:3].T, index=[n.split('__')[1] for n in names], columns=['PC1', 'PC2', 'PC3'])
display(pd.concat({pc: load[pc].abs().sort_values(ascending=False).head(6).index.to_series(index=range(6)) for pc in load}, axis=1))
```

```python
pca_rows = [cv_eval(make_pipeline(preprocess(FEATURES), PCA(n_components=v, random_state=SEED), rf()), FEATURES, f'RF + PCA ({v})')
            for v in [0.6, 0.8, 0.95]]
pca_rows.append(cv_eval(make_pipeline(preprocess(FEATURES, scale=False), rf()), FEATURES, 'RF without PCA'))
pca_rows.append(cv_eval(make_pipeline(preprocess(FEATURES), PCA(0.8, random_state=SEED), LogisticRegression(C=0.05, class_weight='balanced', max_iter=5000)), FEATURES, 'Logistic + PCA (0.8)'))
pca_cmp = pd.DataFrame(pca_rows); display(pca_cmp.round(3))
```

**PCA findings:**
* The predictor space has little shared structure. PC1 and PC2 explain only a small share of the variance, many components are needed to reach 80%, and the two groups overlap completely on PC1–PC2. PC1 is mostly driven by the positively worded STAI items (happy, self-confident, satisfied, secure), i.e. *how* students answer that block. That is not the same as depression risk.
* **PCA in the model reduces performance.** The variance it keeps is not the variance that predicts depression, and random forests already handle many correlated inputs well. **The final model does not use PCA.** This is a legitimate, reportable result: PCA was evaluated properly inside cross-validation and rejected.

## 8. Model comparison and tuning (training CV)

```python
models = {
    'Baseline (always "no")':  make_pipeline(preprocess(FEATURES), DummyClassifier(strategy='prior')),
    'Logistic regression (C=0.01)': make_pipeline(preprocess(FEATURES), LogisticRegression(C=0.01, class_weight='balanced', max_iter=5000)),
    'Logistic regression (C=0.1)':  make_pipeline(preprocess(FEATURES), LogisticRegression(C=0.1, class_weight='balanced', max_iter=5000)),
    'Decision tree (depth 4)': make_pipeline(preprocess(FEATURES, False), DecisionTreeClassifier(max_depth=4, min_samples_leaf=20, class_weight='balanced', random_state=SEED)),
    'Gradient boosting':       make_pipeline(preprocess(FEATURES, False), HistGradientBoostingClassifier(learning_rate=0.03, max_depth=4, max_iter=400, l2_regularization=1.0, class_weight='balanced', random_state=SEED)),
    'Extra trees':             make_pipeline(preprocess(FEATURES, False), ExtraTreesClassifier(300, min_samples_leaf=5, class_weight='balanced_subsample', random_state=SEED)),
}
for leaf in [1, 3, 5, 10]:
    models[f'Random forest (leaf={leaf})'] = make_pipeline(preprocess(FEATURES, False), rf(leaf=leaf))
# Trees don't need one-hot encoding: raw integer codes give a much simpler model for the phone.
models['Random forest (leaf=5, raw codes)'] = rf(leaf=5)

t0 = time.time()
mc = pd.DataFrame([cv_eval(m, FEATURES, k) for k, m in models.items()]).sort_values('cv_pr_auc', ascending=False)
mc.to_csv(ROOT / 'results/model_comparison_cv.csv', index=False)
display(mc.round(3)); print(f'{time.time()-t0:.0f}s')
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.errorbar(mc.cv_roc_auc, mc.name, xerr=mc.cv_roc_auc_sd, fmt='o', label='ROC-AUC')
ax.errorbar(mc.cv_pr_auc, mc.name, xerr=mc.cv_pr_auc_sd, fmt='s', label='PR-AUC')
ax.invert_yaxis(); ax.legend(); ax.set(title='Model comparison — repeated 5-fold CV on training set', xlabel='score')
savefig('08_model_comparison')
```

**Decision:** Random forests do best. They combine many weak, non-linear item signals, which the EDA said would be needed. Leaf size 3–10 performs about the same. We choose **`min_samples_leaf=5` on raw integer codes**: it matches the one-hot version, and it means the phone app only has to pass the answers through as numbers, with no encoding step. That makes the mobile export simple and less error-prone.

## 9. Compact question set for the app

The full set has 66 inputs (25 background/phone-habit fields, the TMD total and 40 STAI items), which is a long form. We test shorter forms. **Feature selection happens inside each CV fold**, because selecting on all training data and then cross-validating would leak information and overstate the score.

```python
class TopKByForest(BaseEstimator, TransformerMixin):
    "Keep the k inputs with the highest random-forest importance, learned on the training fold only."
    def __init__(self, k=20): self.k = k
    def fit(self, X, y):
        m = RandomForestClassifier(300, min_samples_leaf=5, class_weight='balanced_subsample', random_state=SEED, n_jobs=1).fit(X, y)
        self.idx_ = np.argsort(m.feature_importances_)[::-1][:self.k]; return self
    def transform(self, X): return np.asarray(X)[:, self.idx_]

t0 = time.time()
ks = [10, 15, 20, 25, 30, 40]
sel = pd.DataFrame([cv_eval(make_pipeline(TopKByForest(k), rf()), FEATURES, f'Top {k} inputs', k=k) for k in ks] +
                   [cv_eval(rf(), FEATURES, f'All {len(FEATURES)} inputs', k=len(FEATURES))]).drop(columns='n_inputs')
sel.to_csv(ROOT / 'results/compact_form_selection_cv.csv', index=False)
display(sel.round(3)); print(f'{time.time()-t0:.0f}s')
fig, ax = plt.subplots(figsize=(8, 3.5))
ax.errorbar(sel.k, sel.cv_pr_auc, yerr=sel.cv_pr_auc_sd, fmt='o-', label='PR-AUC'); ax.errorbar(sel.k, sel.cv_roc_auc, yerr=sel.cv_roc_auc_sd, fmt='s-', label='ROC-AUC')
ax.set(xlabel='number of app inputs', title='Accuracy vs questionnaire length (nested selection)'); ax.legend()
savefig('09_compact_question_set')
```

```python
# Choose the smallest k within ~0.01 PR-AUC of the best, then fix the field list using the whole training set.
best = sel.cv_pr_auc.max()
K = int(sel[sel.cv_pr_auc >= best - 0.01].k.min())
selector = TopKByForest(K).fit(train_df[FEATURES], ytr)
APP_FIELDS = [FEATURES[i] for i in selector.idx_]
# The TMD total needs all 22 TMD answers in the app; keep it only if selected.
print(f'Chosen compact form: {K} inputs')
display(pd.DataFrame({'field': APP_FIELDS, 'question': [lab.get(int(f.split('_')[-1]), '') if f.startswith('STAI') else '' for f in APP_FIELDS]}))
```

## 10. Final model, decision threshold and the one-time test evaluation

**Threshold.** The default cut-off of 0.5 is arbitrary. For a *screening* app, missing a student who is struggling (false negative) costs more than an unnecessary suggestion to check in (false positive). We choose the threshold from **out-of-fold training predictions** so that sensitivity (recall) is about 75%, and then apply it unchanged to the test set.

The app shows two levels: **Low** and **Elevated**. A third "High" band was tried, but too few students fell into it to estimate it reliably, so it was dropped.

```python
final_model = rf(n=300, leaf=5)
oof = cross_val_predict(final_model, train_df[APP_FIELDS], ytr, cv=StratifiedKFold(5, shuffle=True, random_state=SEED), method='predict_proba', n_jobs=-1)[:, 1]
prec, rec, thr = precision_recall_curve(ytr, oof)
THRESH_ELEVATED = float(thr[np.where(rec[:-1] >= 0.75)[0].max()])
print(f'OOF training ROC-AUC {roc_auc_score(ytr, oof):.3f}, PR-AUC {average_precision_score(ytr, oof):.3f}')
print(f'Elevated threshold (≈75% training sensitivity) = {THRESH_ELEVATED:.3f}')

final_model.fit(train_df[APP_FIELDS], ytr)
p_te = final_model.predict_proba(test_df[APP_FIELDS])[:, 1]
def metrics(y, p, t):
    yhat = (p >= t).astype(int); tn, fp, fn, tp = confusion_matrix(y, yhat).ravel()
    return {'threshold': round(t, 3), 'ROC-AUC': roc_auc_score(y, p), 'PR-AUC': average_precision_score(y, p),
            'recall (sensitivity)': recall_score(y, yhat), 'specificity': tn / (tn + fp), 'precision': precision_score(y, yhat, zero_division=0),
            'F1': f1_score(y, yhat), 'balanced accuracy': balanced_accuracy_score(y, yhat), 'TP': tp, 'FP': fp, 'FN': fn, 'TN': tn}
test_res = pd.DataFrame({
    'Baseline (always no)': metrics(yte, np.full(len(yte), ytr.mean()), 0.5),
    'Previous feature set (socio + TMD total), RF @0.5': metrics(yte, rf(leaf=5).fit(train_df[groups['2. + TMD total (previous project set)']], ytr).predict_proba(test_df[groups['2. + TMD total (previous project set)']])[:, 1], 0.5),
    f'Final model @0.5': metrics(yte, p_te, 0.5),
    f'Final model @ elevated threshold': metrics(yte, p_te, THRESH_ELEVATED),
}).T
test_res.to_csv(ROOT / 'results/test_results.csv')
display(test_res.astype(float).round(3))
```

```python
fig, ax = plt.subplots(1, 3, figsize=(16, 4.3))
fpr, tpr, _ = roc_curve(yte, p_te); ax[0].plot(fpr, tpr, label=f'final model (AUC {roc_auc_score(yte, p_te):.2f})'); ax[0].plot([0, 1], [0, 1], 'k--', lw=.8)
ax[0].set(title='ROC curve — test set', xlabel='false-positive rate', ylabel='true-positive rate'); ax[0].legend()
pr_, rc_, _ = precision_recall_curve(yte, p_te); ax[1].plot(rc_, pr_, label=f'final model (AP {average_precision_score(yte, p_te):.2f})')
ax[1].axhline(yte.mean(), color='crimson', ls='--', label='no-skill'); ax[1].set(title='Precision–recall — test set', xlabel='recall', ylabel='precision'); ax[1].legend()
cm = confusion_matrix(yte, (p_te >= THRESH_ELEVATED).astype(int))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False, ax=ax[2], xticklabels=['pred < 11', 'pred ≥ 11'], yticklabels=['actual < 11', 'actual ≥ 11'])
ax[2].set(title=f'Confusion matrix @ threshold {THRESH_ELEVATED:.2f}')
savefig('10_test_evaluation')

band = np.where(p_te >= THRESH_ELEVATED, 'Elevated', 'Low')
bt = pd.DataFrame({'band': band, 'y': yte, 'beck': test_df[SCORE].values}).groupby('band').agg(students=('y', 'size'), share_beck_11plus=('y', 'mean'), mean_beck=('beck', 'mean')).reindex(['Low', 'Elevated'])
display(bt.round(2))
```

```python
pi = permutation_importance(final_model, test_df[APP_FIELDS], yte, scoring='roc_auc', n_repeats=30, random_state=SEED, n_jobs=-1)
imp = pd.DataFrame({'field': APP_FIELDS, 'importance': pi.importances_mean, 'sd': pi.importances_std}).sort_values('importance', ascending=False)
imp['question'] = [lab.get(int(f.split('_')[-1]), '') if f.startswith('STAI') else '' for f in imp.field]
imp.to_csv(ROOT / 'results/permutation_importance_test.csv', index=False)
fig, ax = plt.subplots(figsize=(8, 5))
ax.barh(imp.field, imp.importance, xerr=imp.sd, color='#55A868'); ax.invert_yaxis()
ax.set(title='Permutation importance on test set (drop in ROC-AUC)', xlabel='Δ ROC-AUC'); savefig('11_permutation_importance')
```

**How to read the test results.** On data it has never seen, the final model ranks a struggling student above a non-struggling one about three times out of four (ROC-AUC ≈ 0.75). At the screening threshold it catches about 3 in 4 of the Beck ≥ 11 students. Students in the "Elevated" group are several times more likely to score ≥ 11 than those in "Low". It beats both the baseline and the earlier project model by a clear margin. Precision is moderate, so many flagged students are false alarms. That is acceptable for a gentle "consider checking in with someone" message, but **not** for a diagnosis.

With only about 44 positive students in the test set, these figures carry noticeable uncertainty (± several points). The repeated-CV mean and SD from Sections 8–9 are the more stable estimate.

## 11. Predicting the Beck score itself (regression)

The original aim was to predict the Beck *score*. We train a random-forest regressor on the same inputs and compare it to always predicting the training median. Because the score is so skewed, we report MAE and R², and we also check how well the predicted score *ranks* students for the ≥ 11 decision.

```python
reg = RandomForestRegressor(n_estimators=300, min_samples_leaf=5, max_features='sqrt', random_state=SEED, n_jobs=-1)
cvr = cross_validate(reg, train_df[APP_FIELDS], train_df[SCORE], cv=CV, scoring=['neg_mean_absolute_error', 'r2'], n_jobs=-1)
reg.fit(train_df[APP_FIELDS], train_df[SCORE]); s_te = reg.predict(test_df[APP_FIELDS])
med = np.median(train_df[SCORE])
reg_res = pd.DataFrame({
    'CV (train)': {'MAE': -cvr['test_neg_mean_absolute_error'].mean(), 'R2': cvr['test_r2'].mean()},
    'Test': {'MAE': mean_absolute_error(test_df[SCORE], s_te), 'R2': r2_score(test_df[SCORE], s_te)},
    'Test baseline (median)': {'MAE': mean_absolute_error(test_df[SCORE], np.full(len(test_df), med)), 'R2': r2_score(test_df[SCORE], np.full(len(test_df), med))}}).T
display(reg_res.round(3))
print(f'Ranking students by predicted score: test ROC-AUC for Beck >= 11 = {roc_auc_score(yte, s_te):.3f} (classifier: {roc_auc_score(yte, p_te):.3f})')
fig, ax = plt.subplots(figsize=(6, 5)); ax.scatter(test_df[SCORE], s_te, s=12, alpha=.6); ax.plot([0, 48], [0, 48], 'k--', lw=.8)
ax.set(xlabel='actual Beck total', ylabel='predicted', title='Regression — test set'); savefig('12_regression')
reg_res.to_csv(ROOT / 'results/regression_results.csv')
```

**Finding:** The regressor beats the median baseline, but R² is low (about 0.2 in CV, about 0.15 on test). It pulls every prediction towards the middle of the range, so a student who scores 30 might be predicted 10. Reporting an exact score in the app would look precise when it isn't. **The app therefore reports a risk level (Low / Elevated) from the classifier.** The regression result stays in the report as evidence for that choice.

## 12. Export the model for the mobile app

Two deliverables:
1. `models/mindtrack_rf.joblib` is the scikit-learn model, for Python or a server.
2. `app/mindtrack_model.json` + `MindTrackModel.kt` (in the `MindTrack_mobile_app` repository) run the **same random forest on the phone, offline**, with no server and no extra libraries. Each tree is stored as arrays (`feature`, `threshold`, `left`, `right`, `value`). The Kotlin code walks each tree (go left if `x <= threshold`) and averages the leaf probabilities, exactly as scikit-learn does.

`app/test_vectors.json` contains test inputs with the Python probabilities, so the Android unit test can prove the Kotlin port gives identical answers.

```python
def field_meta(f):
    if f.startswith('STAI_item_'):
        i = int(f.split('_')[-1])
        scale = ['Not at all', 'Somewhat', 'Moderately so', 'Very much so'] if i <= 20 else ['Almost never', 'Sometimes', 'Often', 'Almost always']
        return {'question': lab[i], 'options': {str(k + 1): v for k, v in enumerate(scale)}}
    if f == 'TMD_Total':
        return {'question': 'Sum of the 22 Test of Mobile Dependence answers (each 0-4)', 'min': 0, 'max': 88}
    if f in ('Age', 'Semester'):
        return {'question': f, 'min': valid[f][0], 'max': valid[f][1]}
    if f.startswith('App_'):
        return {'question': f'Do you use {f[4:].replace("_", " ").lower()} apps most often?', 'options': {'0': 'No', '1': 'Yes'}}
    return {'question': f, 'allowed_codes': list(range(valid[f][0], valid[f][1] + 1)) if f in valid else None}

trees = []
for est in final_model.estimators_:
    t = est.tree_
    v = t.value[:, 0, :]; v = v / v.sum(axis=1, keepdims=True)
    trees.append({'feature': t.feature.tolist(), 'threshold': [round(float(x), 6) for x in t.threshold],
                  'left': t.children_left.tolist(), 'right': t.children_right.tolist(), 'value': [round(float(x), 8) for x in v[:, 1]]})
export = {'model': 'MindTrack random forest — estimated risk that Beck total >= 11', 'version': '2.0', 'sklearn': __import__('sklearn').__version__,
          'features': APP_FIELDS, 'field_info': {f: field_meta(f) for f in APP_FIELDS},
          'thresholds': {'elevated': round(THRESH_ELEVATED, 6)},
          'disclaimer': 'Screening estimate from questionnaire answers. Not a medical diagnosis.',
          'trees': trees}
(ROOT / 'app/mindtrack_model.json').write_text(json.dumps(export, separators=(',', ':')))
joblib.dump({'model': final_model, 'features': APP_FIELDS, 'thresholds': export['thresholds']}, ROOT / 'models/mindtrack_rf.joblib')

# Parity check: pure-Python re-implementation of the JSON vs scikit-learn
def predict_json(model, row):
    tot = 0.0
    for t in model['trees']:
        n = 0
        while t['left'][n] != -1:
            n = t['left'][n] if row[t['feature'][n]] <= t['threshold'][n] else t['right'][n]
        tot += t['value'][n]
    return tot / len(model['trees'])
mj = json.loads((ROOT / 'app/mindtrack_model.json').read_text())
pj = np.array([predict_json(mj, r) for r in test_df[APP_FIELDS].to_numpy(float)])
print(f"JSON size: {(ROOT / 'app/mindtrack_model.json').stat().st_size/1e6:.2f} MB | max |JSON - sklearn| on test set = {np.abs(pj - p_te).max():.2e}")
assert np.abs(pj - p_te).max() < 1e-5

vec = [{'inputs': dict(zip(APP_FIELDS, map(float, r))), 'expected_probability': float(p)} for r, p in zip(test_df[APP_FIELDS].to_numpy()[:20], p_te[:20])]
(ROOT / 'app/test_vectors.json').write_text(json.dumps(vec, indent=1))
pd.DataFrame([{'order': i, 'field': f, **{k: json.dumps(v) if isinstance(v, (dict, list)) else v for k, v in field_meta(f).items()}} for i, f in enumerate(APP_FIELDS)]).to_csv(ROOT / 'app/app_input_fields.csv', index=False)
print('Saved: models/mindtrack_rf.joblib, app/mindtrack_model.json, app/test_vectors.json, app/app_input_fields.csv')
```
