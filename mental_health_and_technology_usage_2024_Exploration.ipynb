a. How many rows and columns are contained in the dataset? (3 marks)

```python
#importing the pandas library and calling it pd
import pandas as pd

#store the dataset in an object called df
df = pd.read_csv("mental_health_and_technology_usage_2024.csv")

number_of_rows, number_of_columns = df.shape # the .shape tells you the size of the table
#prints the number of rows and columns
print(f"Number of rows: {number_of_rows}")
print(f"Number of columns: {number_of_columns}")
```

c. Is the dataset complete? i.e. no missing values? (2 marks)

```python
"""
check for missing values in each column, isnull() checks every cell in the dataset and identifies if it is null.
the sum() function is then used to get the total number of missing values in the entire dataset.
"""
missing_values = df.isnull().sum()
print(missing_values)
"""
if the total number of missing values is zero, it means there are no missing values in the dataset. Otherwise, there are missing values that need to be addressed.
"""
if missing_values.sum() == 0:
    print("No missing values found in the dataset.")
else:
    print("Missing values detected in the dataset.")
```
